```python
a= int(input("Enter your grade"))
if a>=91 and a<=100:
    print("A")
elif a>=81 and a<=90:
    print("B")
elif a>=71 and a<=80:
    print("C")
elif a>=61 and a<=70:
    print("D")
elif a>=51 and a<=60:
    print("E")
else:
    print("F")
```

# Task 2

```python
n = int(input("Enter a number"))
sum_even=0
loopp=0

for i in range(1,n+1):
    if i%2==0:
        sum_even+=i
    if i%3==0:
        loopp+=1

    if i%3==0 and i%5==0:
        print("3 5")
    elif i%3==0:
        print("3")
    elif i%5==0:
        print("5")
    else:
        print(i)

print(f"{loopp} numbers are divisible by 3 ")
```
